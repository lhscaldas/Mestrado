```python
import pandas as pd
import os
import matplotlib.pyplot as plt

# Carregamento do dataset
folder = "NDT dataset"
ndt_file = os.path.join(folder, 'NDT_raw.csv')
df_ndt = pd.read_csv(ndt_file)
client_file = os.path.join(folder, 'devices.csv')
df_clients = pd.read_csv(client_file)

# Listar colunas em cada DataFrame
print("Colunas do DataFrame NDT:")
print(df_ndt.columns)
print("\nColunas do DataFrame Clients:")
print(df_clients.columns)
```

```python
# Contar mac_address únicos em df_ndt e mac em df_clients
unique_mac_ndt = df_ndt['mac_address'].nunique()
unique_mac_clients = df_clients['mac'].nunique()
print(f"\nNúmero de mac_address únicos em df_ndt: {unique_mac_ndt}")
print(f"Número de mac únicos em df_clients: {unique_mac_clients}")
```

```python
# printar os macs que existem em df_ndt mas não existem em df_clients
macs_ndt = set(df_ndt['mac_address'].unique())
macs_clients = set(df_clients['mac'].unique())
macs_only_ndt = macs_ndt - macs_clients
print(f"\nNúmero de mac_address únicos em df_ndt que não existem em df_clients: {len(macs_only_ndt)}")
print(f"mac_address únicos em df_ndt que não existem em df_clients: {macs_only_ndt}")
```

```python
client_file_old = os.path.join(folder, 'devices_old.csv')
df_clients_old = pd.read_csv(client_file_old)
# verificar se os macs que existem em df_ndt mas não existem em df_clients_old
macs_clients_old = set(df_clients_old['mac'].unique())
macs_only_ndt_old = macs_only_ndt - macs_clients_old
print(f"\nNúmero de mac_address únicos em df_ndt que não existem em df_clients mas existem em df_clients_old: {len(macs_only_ndt_old)}")
print(f"mac_address únicos em df_ndt que não existem em df_clients mas existem em df_clients_old: {macs_only_ndt_old}")
```

```python
# contar quantas medições em df_ndt são dos macs_only_ndt (total e por mac)
ndt_only_ndt = df_ndt[df_ndt['mac_address'].isin(macs_only_ndt)]
print(f"\nNúmero de medições em df_ndt que são dos macs_only_ndt: {len(ndt_only_ndt)}")
for mac in macs_only_ndt:
    count = len(ndt_only_ndt[ndt_only_ndt['mac_address'] == mac])
    print(f"  - {mac}: {count}")
```
